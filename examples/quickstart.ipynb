# DataPrep quickstart

Explore a DataFrame in a few lines. Every output below is self-contained: nothing is loaded from the internet, so it works on closed networks and inside VS Code notebooks.

```python
from dataprep.datasets import load_dataset
from dataprep.eda import create_report, plot, plot_correlation, plot_diff, plot_missing

df = load_dataset("titanic")
df.head()
```

## Every column at a glance

`plot(df)` draws one chart per column and lists data-quality insights.

```python
plot(df)
```

## One column in depth

```python
plot(df, "Age")
```

## How two columns relate

```python
plot(df, "Age", "Sex")
```

## Correlations

```python
plot_correlation(df)
```

## Missing values

```python
plot_missing(df)
```

## Compare two DataFrames

```python
plot_diff([df[df["Survived"] == 1], df[df["Survived"] == 0]], config={"diff.label": ["survived", "died"]})
```

## Which columns relate to a target?

`plot(df, target=...)` ranks every column by its association with the target (0 = none, 1 = perfect). The **!** button lists warnings such as ID-like columns or possible target leakage.

```python
plot(df, target="Survived")
```

## Statistics as a DataFrame

`summarize(df)` returns the numbers behind the plots, one row per column, for filtering, saving or checks in a pipeline.

```python
from dataprep.eda import rank_features, summarize

summary = summarize(df, target="Survived")
summary
```

```python
summary.loc[summary["missing_pct"] > 5, ["missing", "missing_pct", "insights"]]
```

## Large tables

Every function accepts `sample=`: a number of rows or a fraction. The output says that it shows a sample.

```python
plot(df, sample=500)
```

## A full profiling report

The report is one HTML document. `report.save("titanic.html")` writes it to disk.

```python
report = create_report(df, title="Titanic")
report
```
